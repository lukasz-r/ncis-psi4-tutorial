```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sto_1s(r, Z=1):
    return (Z**1.5 / np.sqrt(np.pi)) * np.exp(-Z * r)

    
    # Define Gaussian-type orbital (STO-3G approximation)
def gto_1s(r, alpha=[3.42525091, 0.62391373, 0.16885540], coeff=[0.15432897, 0.53532814, 0.44463454]):
    result = np.zeros_like(r)
    for c, a in zip(coeff, alpha):
        result += c * (2 * a / np.pi)**(3/4) * np.exp(-a * r**2)
    return result
```

```python
R_AB = 5.0
```

```python
x = np.linspace(-5, 8, 500)
```

```python
r_A = np.abs(x)
r_B = np.abs(x - R_AB)
```

```python
psi_A = sto_1s(r_A)
psi_B = sto_1s(r_B)
overlap = psi_A * psi_B
```

```python
psi_A_g = gto_1s(r_A)
psi_B_g = gto_1s(r_B)
overlap_g = psi_A_g * psi_B_g
```

```python
plt.figure(figsize=(8, 4))
plt.plot(x, psi_A, label='Wavefunction Atom A')
plt.plot(x, psi_B, label='Wavefunction Atom B')
plt.plot(x, overlap*50, label='Overlap (Product) x 10', linestyle='--', color='black')
plt.axvline(0, color='gray', linestyle=':', linewidth=0.8)
plt.axvline(R_AB, color='gray', linestyle=':', linewidth=0.8)

plt.xlabel('Position (Bohr)')
plt.ylabel('Wavefunction amplitude')
plt.legend()
plt.title('Overlap of Two Hydrogen Atom Wavefunctions')
plt.grid(True)
#plt.show()

#plt.figure(figsize=(8, 4))
plt.plot(x, psi_A_g, label='GTO Atom A')
plt.plot(x, psi_B_g, label='GTO Atom B')
plt.plot(x, overlap_g*50, label='GTO Overlap (Product) x 10', linestyle='--', color='black')
plt.axvline(0, color='gray', linestyle=':', linewidth=0.8)
plt.axvline(R_AB, color='gray', linestyle=':', linewidth=0.8)

plt.xlabel('Position (Bohr)')
plt.ylabel('Wavefunction amplitude')
plt.legend()
#plt.title('Overlap of Two GTOs ')
#plt.grid(True)
plt.show()
```
